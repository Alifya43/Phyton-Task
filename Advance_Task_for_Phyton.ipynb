```python
#Task 11: Peersonal Expense Tracker
total_expense = int(input("How many expenses do you want to add: "))

expenses_list = []

for i in range(1, total_expense + 1):
  name = input("Enter your expense name: ")
  category = input("Enter your expense category: ")
  amount = float(input("Enter your expense amount: "))

  expense_item = {
      "expense_name" : name,
      "category" : category,
      "amount": amount
  }

  expenses_list.append(expense_item)

total_spending = 0
for expense_item in expenses_list:
  total_spending = total_spending + expense_item["amount"]

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping": 0,
    "Bills": 0,
    "Other": 0
}

for expense_item in expenses_list:
    category = expense_item["category"]
    amount = expense_item["amount"]

    if category in category_spending:
        category_spending[category] = category_spending[category] + amount
    else:
        category_spending["Other"] = category_spending["Other"] + amount

highest_expense = expenses_list[0]
for expense_item in expenses_list:

    if expense_item["amount"] > highest_expense["amount"]:
        highest_expense = expense_item

budget = float(input("Enter your budget: "))

if total_spending > budget:
    print("Warning! Budget exceeded.")
else:
    print("You are within your budget.")

def final_report():

    print("\n----- Expense Report -----")

    print("Total Spending:", total_spending)

    print("\nCategory Spending")

    for category, amount in category_spending.items():
        print(category, ":", amount)

    print("\nHighest Expense")
    print("Name:", highest_expense["expense_name"])
    print("Amount:", highest_expense["amount"])

final_report()





```

```python
#Task 12: Store Inventory Manager
products={}

while True:
  print("========Products========")
  print("1.Add Product")
  print("2.Sell Products")
  print("3.Restock Products")
  print("4.Search products")
  print("5.Display Inventory")
  print("6.Exit")

  choice=int(input("Enter your choice:"))
  if choice==1:
    name=input("Enter product name:")
    price=float(input("Enter product price:"))
    quantity=int(input("Enter product quantity:"))
    products[name]={"price":price,"quantity":quantity}
    print("Product added successfully")
  elif choice ==2:
    name=input("Enter product name:")
    if name in products:
      quantity=int(input("Enter quantity sold:"))
      if quantity<=products[name]["quantity"]:
        products[name]["quantity"]-=quantity
        print("Products sold successfully")
      else:
        print("Insufficient quantity")
    else:
      print("Product not found")

  elif choice==3:
    name=input("Enter product name:")
    if name in products:
      quantity=int(input("Enter quantity to restock:"))
      products[name]["quantity"]+=quantity
      print("Products restocked successfully")
    else:
      print("Product not found")
  elif choice==4:
    name=input("Enter Product name: ")
    if name in products:
      print("Product found")
      print("Name:",name)
      print("Price:",products[name]["price"])
      print("Quantity:",products[name]["quantity"])
    else:
      print("Product not found")

  elif choice == 5:
        print("Inventory:")

        total_value = 0

        for name, info in products.items():
            print("Name:", name)
            print("Price:", info["price"])
            print("Quantity:", info["quantity"])

            if info["quantity"] < 5:
                print("Low Stock")

            product_value = info["price"] * info["quantity"]
            total_value += product_value

            print()

        print("Total Inventory Value:", total_value)

  elif choice == 6:
     print("Thank you!")
     break

  else:
        print("Invalid choice")
```

```python
#Task 13: Library Book Management System
books = []


def add_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists")
            return

    title = input("Enter book title: ")
    author = input("Enter book author: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "status": "Available"
    }

    books.append(book)
    print("Book added successfully")


def display_books():
    if len(books) == 0:
        print("No books available")
        return

    print("\nBooks:")

    for book in books:
        print("ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", book["status"])
        print()


def search_book():
    title = input("Enter book title to search: ")

    for book in books:
        if book["title"].lower() == title.lower():
            print("Book found")
            print("ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])
            print("Status:", book["status"])
            return

    print("Book not found")


def borrow_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Available":
                book["status"] = "Borrowed"
                print("Book borrowed successfully")
            else:
                print("Book is not available")

            return

    print("Book not found")


def return_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:

            if book["status"] == "Borrowed":
                book["status"] = "Available"
                print("Book returned successfully")
            else:
                print("Book is already available")

            return

    print("Book not found")


def menu():
    while True:
        print("\n1. Add Book")
        print("2. Display Books")
        print("3. Search Book")
        print("4. Borrow Book")
        print("5. Return Book")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            add_book()

        elif choice == 2:
            display_books()

        elif choice == 3:
            search_book()

        elif choice == 4:
            borrow_book()

        elif choice == 5:
            return_book()

        elif choice == 6:
            print("Thank you!")
            break

        else:
            print("Invalid choice")


menu()
```

```python
# Task 14: Restaurant Ordering System

menu= {
    "Burger":500,
    "Pizza":1200,
    "Pasta":800,
    "Drink":200
}

orders = []

print("========= Restaurant Menu =========")
for key,value in menu.items():
  print(f'{key} : {value}')

while True:
  item_name = input("Enter your item name: ").title()
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
    print("This item is not available")

  choice = input("You want to add more items: yes/no: ")
  if choice == "no":
    break

total = 0
for order in orders:
  total = total + order["sub_total"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1
else:
  discount = 0

grand_total = total + tax - discount

def receipt():
  print("\n============ Receipt ============")

  for order in orders:
    print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')

  print("Total: ", total)
  print("Tax: ", tax)
  print("Discount: ", discount)
  print("Grand Total: ", grand_total)

receipt()
```

```python
# Task 15: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient found")
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')
      print(f'Patient emergency: {patient["emergency"]}')
      return
    else:
      print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted")

def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("Invalid choice")
```
