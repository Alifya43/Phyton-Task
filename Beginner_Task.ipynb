```python
#1:cinema ticket price
name=input("Enter Name:")
age=int(input("Enter Age:"))

if age < 5 or age > 100:
    price = 0

elif age >= 5 and age <= 12:
    price = 300

elif age >= 13 and age <= 59:
    price = 700

elif age >= 60 and age <= 100:
    price = 400

print("Customer:", name)
print("Ticket Price: Rs.", price)



```

```python
#2:Temperature Converter
celcius_temperature=int(input("Enter Temp: "))
fahrenheit=(celcius_temperature*9/5)+32
print("Temp in Fahrenheit:",fahrenheit)

if celcius_temperature<10:
  print("Very Cold")
elif celcius_temperature>=10 and celcius_temperature<=19:
  print("Cold")
elif celcius_temperature>=20 and celcius_temperature<=29:
  print("Normal")
else:
  print("Hot")
```

```python
#3:Mobile Recharge Calculator
mobile_num=int(input("Enter Mobile Number: "))
recharge_amount=int(input("Enter Recharge Amount"))
company_tax=recharge_amount*3/100
final_balance= recharge_amount-company_tax
print("=========Mobile Recharge=========")
print("Mobile Number:",mobile_num)
print("Recharge Amount:",recharge_amount)
print("Company Tax:",company_tax)
print("Final Balance:",final_balance)
```

```python
#4:Weekly Steps Tracker
total=0
for i in range(7):
  daily_steps=int(input("Enter Steps for Day:"+str(i+1)+":"))
  total=total+daily_steps

average=int(total/7)

print("Total Steps:",total)
print("Daily Average:",average)

if average>=5000:
  print("Status:Active Week")
else:
  print("Status:Need More Activity")
```

```python
#5:Secure PIN Checker
secret_pin=1234

for i in range(3):
  enter_pin=int(input("Enter Pin:"))
  if enter_pin==secret_pin:
    print("Access Granted")
    break
  else:
    remaining=2-i
    print("Incorrect Pin")
    print("Access Denied.Remaining Attempts:",remaining)
else:
    print("XXXACCOUNT BLOCKEDXXX")




```
